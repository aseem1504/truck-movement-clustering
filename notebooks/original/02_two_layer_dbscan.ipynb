```python
from sklearn.cluster import DBSCAN
```

```python
layer1=pd.read_csv("movement_similarity_eps10_minsamp_10.csv")
```

```python
del layer1['Unnamed: 0']
```

```python
for i in layer1.drop(['vehicle_number','DB_clustered'],axis=1).columns:
    if layer1[i].sum()==0:
        layer1.drop(i,axis=1,inplace=True)
```

```python
len(layer1.columns)
```

```python
dict_of_dataframes={}
for i in layer1.DB_clustered.unique().tolist():
    reuse=layer1[layer1['DB_clustered']==i]
    dict_of_dataframes.update({'cluster_{}'.format(i):reuse})
    
```

```python
for j in dict_of_dataframes.keys():
    for i in dict_of_dataframes[j].drop(['vehicle_number','DB_clustered'],axis=1).columns:
        if dict_of_dataframes[j][i].sum()==0:
            dict_of_dataframes[j].drop(i,axis=1,inplace=True)
            
```

```python
for j in dict_of_dataframes.keys():
    print(len(dict_of_dataframes[j].drop(['vehicle_number','DB_clustered'],axis=1).columns))
```

```python
for j in dict_of_dataframes.keys():
    x=dict_of_dataframes[j].drop(['vehicle_number','DB_clustered'],axis=1).values
    clustering = DBSCAN(eps=2, min_samples=5,n_jobs=-1).fit(x)
    dict_of_dataframes[j].loc[:,'DB_clustered_layer2']=clustering.labels_
    
```

```python
two_layer_clustering=pd.DataFrame()
for j in dict_of_dataframes.keys():
    reuse=dict_of_dataframes[j][['vehicle_number','DB_clustered','DB_clustered_layer2']]
    two_layer_clustering=pd.concat([reuse,two_layer_clustering])
    
```

```python
final_clustered_data=pd.merge(layer1,two_layer_clustering,on='vehicle_number',how='left')
```
